## Algorithm: Simple Moving Average (SMA)

Now that we have an idea of what our data will look like how will we analyse it? 
For this exercise we'll be using Simple Moving Average (SMA). 
Simple Moving Average (SMA) is an ideal lead-in algorithm for commodity trading. 
Commodity buyers and traders at coffee roasters use SMAs to average or smooth out day-to-day spot price noise (or volatility) and detect underlying supply chain pricing trends.

The SMA is calculated by adding the closing prices for the last `k` periods and dividing by `k`. The concise form of the formula for a k-period moving average at time t is:
$$
\text{SMA}_t = \frac{1}{k} \sum_{i=0}^{k-1} P_{t-i}
$$

We can expand the expression to illustrate the calculation of the arithmetic mean of an asset’s closing prices over the rolling period `k`. Every price in the window receives equal weight (unlike Exponential Moving Averages, which weight more recent prices more heavily). 
$$
\text{SMA}_t = \frac{P_t + P_{t-1} + P_{t-2} + \dots + P_{t-k+1}}{k} = \frac{1}{k} \sum_{i=0}^{k-1} P_{t-i}
$$

## Pseudo-code: Simple Moving Average (SMA)

Programmers use pseudo-code to transition from an abstract algorithm to hand-written code.
By sketching out loop conditions, array bounds, and intermediate variables in plain language first, students learn to separate algorithmic logic from language-specific syntax errors.
Let's consider the following mock `list` of closing prices from January 2020: 5 daily closing spot prices for Arabica coffee (in cents/lb):

Prices=[127.2,128.2,129.2,130.2,131.2]

Let's calculate a 3-day SMA (k=3) by hand on paper:

Day 1 SMA: meaningless, as we need at least 3 days of data to calculate a 3-day SMA.
Day 2 SMA: meaningless, as we need at least 3 days of data to calculate a 3-day SMA.
Day 3 SMA: (127.2 + 128.2 + 129.2) / 3
Day 4 SMA: (128.2 + 129.2 + 130.2) / 3
Day 5 SMA: (129.2 + 130.2 + 131.2) / 3
​	
Pseudocode is the bridge that links our understanding of the problem to its implementation.
We must force ourselves to identify required inputs, calculate loop boundaries, and plan sliding window mechanics explicitly.

```plain-text
# Pseudo-code for calculating a k-day SMA
1. Define a function to calculate SMA with parameters: prices (list of closing prices) and k (number of days for the moving average)
2. Initialize an empty list to store SMA values
3. Loop through the prices list starting from index k-1 to the end of the list 
    a. For each index i, calculate the sum of the last k prices (from i-k+1 to i)
    b. Divide the sum by k to get the SMA value
    c. Append the SMA value to the corresponding row in the SMA list
4. Return the list of SMA values
```


## Hand-Written Python Function

Let's translate our pseudocode line-by-line into a working Python function using only native data structures and standard loops (without NumPy or external modules):

```python
def calculate_sma(prices, k) -> list:
    sma_values = []  # Initialize an empty list to store SMA values
    for i in range(k - 1, len(prices)):  # Loop through the prices list starting from index k-1 to the end of the list
        window_sum = sum(prices[i - k + 1:i + 1])  # Calculate the sum of the last k prices (from i-k+1 to i)
        sma = window_sum / k  # Divide the sum by k to get the SMA value
        sma_values.append(sma)  # Append the SMA value to the corresponding row in the SMA list
    return sma_values
price_list = [127.2, 128.2, 129.2, 130.2, 131.2]
print(calculate_sma(price_list, 3))  # Example usage
```

## The NumPy Vectorized 
This final step demonstrates how NumPy can be used to replace explicitly written Python loops with a single vectorized operation (np.convolve or 2D sliding windows) and produce a new numpy array.

```python
import numpy as np
def calculate_sma_numpy(prices, k) -> np.ndarray:
    prices_array = np.array(prices)  # Convert the list of prices to a NumPy array
    """Vectorized 1D convolution replacing explicit Python loops."""
    sma_values = np.convolve(prices_array, np.ones(k)/k, mode='valid')  # Use np.convolve to calculate the SMA values
    return sma_values

print(calculate_sma_numpy(price_list, 3))  # Example usage
# print type of result
print(type(calculate_sma_numpy(price_list, 3)))
```

````python
import numpy as np
def calculate_sma_numpy(prices, k) -> np.ndarray:
    prices_array = np.array(prices)  # Convert the list of prices to a NumPy array
    """Vectorized 1D convolution replacing explicit Python loops."""
    sma_values = np.convolve(prices_array, np.ones(k)/k, mode='valid')  # Use np.convolve to calculate the SMA values
    return sma_values

print(calculate_sma_numpy(price_list, 3))  # Example usage
# print type of result
print(type(calculate_sma_numpy(price_list, 3)))
````
