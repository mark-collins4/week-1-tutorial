## Why Copilot?

Copilot offers dimmed ghost text suggestions as you type: sometimes the completion of the current line, sometimes a whole new block of code. You can accept the suggestion by pressing the Tab key, or you can ignore it and continue typing.

To partially accept a suggestion, use the ⌘→ keyboard shortcut to accept either the next word of a suggestion, or the next line.

Copilot can also generate suggestions from code comments
Instead of relying on Copilot to provide line-by-line suggestions, you can provide hints about what code you expect by using code comments. For example, you could specify a type of algorithm or concept to use (for example, "use recursion" or "use a singleton pattern"), or which methods and properties to add to a class.

### Copilot GOOD or BAD?

Why turn off Copilot? Because you won't learn how to program if you just accept the suggestions without thinking about them. You need to think about what you are doing and why, and then write the code yourself. Copilot can be a useful tool, but it should not be relied upon to do the work for you.

### TWO WAYS TO TURN OFF/ON COPILOT

The first way: To enable or disable inline suggestions, select the Copilot menu in the bottom Status Bar, and then check or uncheck the options to enable or disable inline suggestions. 

The second way: Use the command palette at the top of the editor and search for “Toggle “Hide AI Features” to enable/disable it all.

1. Open the command palette (cmd-shift-p on Mac)
1. Search for “Chat”
1. Select “Chat: Hide AI Features

For more information, see:
* [Toggle AI powered suggestions in Visual Studio Code](https://code.visualstudio.com/docs/editing/ai-powered-suggestions)
