## Data Cleaning and Preprocessing

Data cleaning involves handling missing values (NaN), fixing invalid formats, and ensuring correct data types.

In df4_imports:
- green_weight_kg has a missing value (NaN).
- order_date is stored as object (string) and needs conversion to a valid date/time format or type. Internally, Pandas stores date and time series as datetime64 objects (offering various computational benefits). Externally, Pandas utilizes Python's standard strftime (string format time) directives to parse and display these values datetime64[^dt64].

How do we handle these issues? We'll explore a few options for handling missing values, and then convert the order_date column to the appropriate datetime type.

```python

# Check missing values per column
print("Missing values count:\n", df4_imports.isna().sum())

# 1. There isn't any hard and fast rule about how to fill missing values, so in this case, we'll simply fill missing numeric values with the median of the respective origin/overall metric
median_weight = df4_imports["green_weight_kg"].median()
df4_imports["green_weight_kg"] = df4_imports["green_weight_kg"].fillna(median_weight)

# 2. When we inspected the DataFrame 'df4_imports' we noticed that the column 'order_date' which should obviously be a datetime Dtype, was in fact 'string' Dtype. The right thing to do is to convert to date column to datetime type. We'll use the Pandas `to_datetime` function to automatically handle the conversion (and any errors e.g. invalid formats) under default mode (i.e. Pandas makes a best guess). 
df4_imports["order_date"] = pd.to_datetime(df4_imports["order_date"])

print("\n--- After Cleaning Missing Data & Data Types ---")
print(df4_imports.info())
print("\nCleaned DataFrame:")
print(df4_imports[["contract_id", "green_weight_kg", "order_date"]])

```

## Mastering Text Control

AKA Tidying up text. Text fields often contain accidental whitespace or inconsistent casing (e.g., '  BeanCo Logistics  ', ' Organic '). Pandas provides vectorised string operations via '.str' to strip unwanted whitespace from string columns, to convert column strings to uppercase, to filter DataFrames using selected strings, and to replace string values.

For example:

```python

# Strip unwanted whitespace from string columns
text_cols = ["importer", "origin_country", "certification"]
for col in text_cols:
    df4_imports[col] = df4_imports[col].str.strip()

# Convert certification strings to uppercase
df4_imports["certification_clean"] = df4_imports["certification"].str.upper()

# String matching: Filter imports that contain 'ORGANIC'
organic_imports = df4_imports[df4_imports["certification_clean"].str.contains("ORGANIC")]

# Replace string values
df4_imports["status"] = df4_imports["status"].str.replace("In Transit", "Shipped")

print("--- Cleaned Text Fields & Organic Shipments ---")
print(organic_imports[["contract_id", "importer", "certification_clean"]])

```


## Mastering Data Control

Data control (aka manipulation) uses a mix of tools, filtering, boolean logic, indexing (.loc, .iloc), for computing derived metrics from or within DataFrames: updating, supplimenting, or creating new columns of data for example.

We can perform simple math (+-*/ etc.) and use more complicated queries. Pandas also provides the '.query()' function for analysing and selecting data in a DataFrame.

Let's enhance the DataFrame with two new metrics

- Convert green weight from kilograms to Metric Tons (1 MT=1000 kg).
- Calculate Total Contract Value (USD) (Weight in kg x FOB Price per kg).

```python

# 1. Derive new financial and weight metric columns, we basically state that the new column exists and so it is added to the DataFrame 
df4_imports["green_weight_mt"] = df4_imports["green_weight_kg"] / 1000.0
df4_imports["total_value_usd"] = df4_imports["green_weight_kg"] * df4_imports["fob_price_usd_kg"]

# 2. Label-based selection via .loc and positional-based via .iloc
print("--- All High Value Shipments (.loc) ---")
print(df4_imports.loc[df4_imports["total_value_usd"] > 50000, ["contract_id", "importer", "total_value_usd"]])

print("\n--- First 3 Rows and First 4 Columns (.iloc) ---")
print(df4_imports.iloc[0:3, 0:4])

print("\n--- First 3 Rows and Last 4 Columns (.iloc) ---")
print(df4_imports.iloc[0:3, -4:])

# 3. Filtering rows using boolean query. Here we create a new DataFrame containing the results of the given .query() expression 
# Target high-value shipments (> $50,000) or delivered status
high_val_shipments = df4_imports.query("total_value_usd > 50000 and status == 'Delivered'")

print("--- High Value Delivered Shipments (.loc) ---")
print(high_val_shipments.iloc[ :, -4:])

```

## Applied Lambda Expressions

In Session 5 we introduced 'Lambda' expressions (aka 'Lambda functions' or 'inline functions') as simple single-line, non-reuseable functions to apply custom row-wise or column-wise algorithms. Unlike Funcitons, a Lambda expression is typically small (a single line of code), unnamed, and essentially not intended for reuse across your codebase. They can take any number of arguments but have only one expression. 

Their signature takes the form of 'lambda parameters: expression' and can be used as arguments themselves within higher order functions.

We'll write 'Lambda' expressions for two reports:
- Tier Assignment: Classify contracts based on weight size (Bulk vs. Niche Micro-lot).
- Roasting Loss Calculation: Calculate green-to-roasted weight loss percentage (typically 12%–20% moisture loss during roasting).

We will use Pandas '.apply()' function to apply an expression along an axis of the DataFrame 

Objects passed to the function are Series objects whose index is either the DataFrame’s index (axis=0) or the DataFrame’s columns (axis=1)

```python

# 1. Apply lambda function across a single column (Volume Classification)
df4_imports["lot_tier"] = df4_imports["green_weight_kg"].apply(
    lambda weight: "Bulk Commodity" if weight >= 15000 else "Specialty Micro-lot"
)

# 2. Row-wise lambda function across multiple columns (axis=1)
# Premium pricing rule: Apply a 5% import duty surcharge if FOB price > $7/kg
df4_imports["duty_adjusted_cost_usd"] = df4_imports.apply(
    lambda row: row["total_value_usd"] * 1.05 if row["fob_price_usd_kg"] > 7.0 else row["total_value_usd"],
    axis=1
)

print("--- Applied Lambda Functions Results ---")
print(df4_imports[["contract_id", "fob_price_usd_kg", "total_value_usd", "lot_tier", "duty_adjusted_cost_usd"]])

```

## Data Aggregation

Next up, exercising Pandas DataFrame's data aggregation functions; '.groupby()' and '.pivot_table()'.
We'll use these aggregation functions to combine individual contract rows into summarized data insights.

```python

# 1. Group by Importer: Total metric tons and average price per kg
importer_summary = df4_imports.groupby("importer", as_index=False).agg(
    total_contracts=("contract_id", "count"),
    total_metric_tons=("green_weight_mt", "sum"),
    avg_fob_usd_kg=("fob_price_usd_kg", "mean"),
    total_expenditure_usd=("total_value_usd", "sum")
)

print("--- Importer Summary (Groupby) ---")
print(importer_summary)

# 2. Multi-dimensional Pivot Table:
# Importer vs Status showing Total Green Weight (kg)
pivot_weight = df4_imports.pivot_table(
    index="importer",
    columns="status",
    values="green_weight_kg",
    aggfunc="sum",
    fill_value=0
)

print("\n--- Importer vs Status Pivot Table ---")
print(pivot_weight)

```

## Data Sorting & Selection

Pandas DataFrames has a suite of useful utility functions for interrogating and reporting on datasets.
The following examples apply 'n.sort_values', 'n.smallest' and 'n.largest' to arrange, and top, and tail a dataset.

In this case, let's identify key market drivers, top suppliers, and extreme quality outliers.

```python

# Sort Dataframe by total value descending
sorted_by_val = df4_imports.sort_values(by="total_value_usd", ascending=False)

# Select top 3 highest value contracts using nlargest
top_3_contracts = df4_imports.nlargest(3, columns="total_value_usd")

# Select top 2 lowest price contracts using nsmallest
cheapest_2_contracts = df4_imports.nsmallest(2, columns="fob_price_usd_kg")

print("--- Top 3 Highest Value Import Contracts ---")
print(top_3_contracts[["contract_id", "importer", "origin_country", "total_value_usd"]])

print("\n--- Top 2 Most Cost-Effective Origins ---")
print(cheapest_2_contracts[["contract_id", "origin_country", "fob_price_usd_kg"]])

```


## Merging, Joining & Concatenation

In real commodity supply chains, shipment data lives separately from Quality Assurance (QA) lab reports. We merge df1_contracts and df2_cupping on contract_id.

```python

# 1. Inner Join contracts metadata with quality cupping scores
contract_qa_df = pd.merge(
    df1_contracts,
    df2_cupping,
    on="contract_id",
    how="inner"
)

# 2. Feature Engineering on merged dataset: Identify Specialty Grade (Score >= 80.0)
contract_qa_df["grade"] = contract_qa_df["cupping_score"].apply(
    lambda score: "Specialty Grade" if score >= 80.0 else "Commercial Grade"
)

print("--- Merged Contract & QA Cupping Data ---")
print(contract_qa_df[["contract_id", "origin_country", "variety", "cupping_score", "moisture_pct", "grade"]])

# 3. Concatenate new incoming lots vertically
new_shipments = pd.DataFrame({
    "contract_id": ["CT-106", "CT-107"],
    "origin_country": ["Kenya", "Costa Rica"],
    "variety": ["SL28", "Villa Sarchi"],
    "bag_count_60kg": [200, 180],
    "fob_price_usd_lb": [5.20, 4.80],
    "shipment_date": ["2026-04-01", "2026-04-05"]
})

all_contracts_df = pd.concat([df1_contracts, new_shipments], ignore_index=True)
print("\n--- Expanded Contracts List (Concatenated) ---")
print(all_contracts_df.tail(4))

```

## What About Series?

Pandas also has a `series` object in addition to `dataframe`. What is it for? When and how is it used? 
The `series` object is a 1-dimensional data structure whereas a DataFrame is 2-dimensional. In fact, a DataFrame can be thought of as a collection of Series objects that share a common index. Think of each column in a DataFrame as a Series. 

A Series holds a single 1D array of data (integers, floats, strings, objects, datetimes) combined with an associated set of row labels called an Index.

A Series represents a single series of data, sequential data like daily temperatures, stock prices, or customer names. Built upon NumPy it allows for fast vectorized operations that allow for easy element-wise calculations (e.g., adding taxes, multiplying columns) without writing explicit for loops.


### How to use a Series

1. Creating a Series

You can create a standalone Series from a Python list, dictionary, or NumPy array:

```python

import pandas as pd

# From a list with custom index labels
prices = pd.Series([3.45, 2.85, 1.95], index=["Ethiopia", "Colombia", "Brazil"], name="fob_price_usd")

# From a dictionary (keys automatically become the index)
cupping_scores = pd.Series({"CT-101": 88.5, "CT-102": 84.0, "CT-103": 81.5})

```

2. Extracting a Series from a DataFrame

Selecting a single column from a DataFrame returns a Series:

```python

# Assuming sales_df exists
revenue = sales_df["revenue"]  # Returns a pandas Series

```

3. Accessing Values

You can access elements by position (.iloc) or by index label (.loc):

```python

print(prices.loc["Ethiopia"])  # Output: 3.45
print(prices.iloc[0])          # Output: 3.45

```

4. Vectorized Arithmetic & Filtering

Operations apply to every element automatically:

```python

# Increase all prices by 10%
prices_with_duty = prices * 1.10

# Filter values using boolean conditions
high_grade_prices = prices[prices > 2.50]

```

5. Common Methods

Series objects come with built-in analytical methods:

```python

# Statistical summaries
mean_price = prices.mean()
max_price = prices.max()

# String manipulations (for text Series)
countries = pd.Series([" ethiopia ", "colombia ", "brazil"])
clean_countries = countries.str.strip().str.title()  # ['Ethiopia', 'Colombia', 'Brazil']

# Categorical counts
status_series = pd.Series(["Delivered", "In Transit", "Delivered", "Pending"])
print(status_series.value_counts())

```
